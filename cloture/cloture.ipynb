# Filibuster and Cloture on Legislation

Cloture motions from the Senate's tables at https://www.senate.gov/legislative/cloture/{congress}.htm (plain HTTP, no browser). See the README for the counting rules. Quarters are assigned by the date each motion was filed. The scraper prints a line for anything unusual on the Senate's page and for any missing vote result it fills in from the official roll call.

```python
import cloture_helper

CONGRESSES = [117, 118, 119]
```

```python
for congress in CONGRESSES:
    results = cloture_helper.scrape_cloture(congress)
    summary = cloture_helper.summarize_cloture(results)
    results.to_csv(f"cloture_results_{congress}.csv", index=False)
    summary.to_csv(f"cloture_summary_{congress}.csv")
    print(f"{congress}th Congress: {len(results)} rows")
    display(cloture_helper.published_rows(summary))
```

The tables above are cumulative within each Congress, the format of the "Cloture Motion Activity" sheet in the full data tables.

`cloture_summary_{congress}.csv` has the per-quarter internal counts (motions filed, votes, invoked, failed, and reconsidered) for Legislation, Nominations, and Motions to Proceed. `cloture_results_{congress}.csv` has one row per cloture motion plus one row per reconsideration vote. Votes are the rows with Result I (invoked) or F (failed).

The Senate's page can lag the floor by several days, so rerun the notebook after a quarter closes.
